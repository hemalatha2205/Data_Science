```python
#Task1
name = input("enter name: ")
age = int(input("enter age: "))
fav_no = float(input("enter favorite number: "))
print(f"hello, {name}!")
current_year = 2025
year_in_100 = current_year + (100 - age)
print(f"you will turn 100 in the year {year_in_100}")
doubled = fav_no * 2
print(f"your favorite number doubled is {doubled}")
```

```python
#Task2
num = int(input("enter a number: "))
if num % 2 == 0:
    print("the number is even")
else:
    print("the number is odd")
if num > 0:
    print("the number is positive")
elif num < 0:
    print("the number is negative")
else:
    print("the number is zero")
```

```python
#Task3
print("even numbers from 1 to 50 using for loop:")
for i in range(2, 51, 2):
    print(i)
print("even numbers from 1 to 50 using while loop:")
i = 2
while i <= 50:
    print(i)
    i += 2
```

```python
#Task4
foods = []
print("enter 5 favorite foods:")
for i in range(5):
    food = input(f"food {i+1}: ")
    foods.append(food)
print("your favorite foods:", foods)
food_lengths = {}
for food in foods:
    food_lengths[food] = len(food)
print("food and name length:", food_lengths)
```

```python
#Task5
def calculate_area(shape, d1, d2=None):
    if shape == "circle":
        return 3.14 * d1 * d1
    elif shape == "rectangle":
        return d1 * d2
    elif shape == "triangle":
        return 0.5 * d1 * d2
    else:
        return None
print("circle area:", calculate_area("circle", 6))
print("rectangle area:", calculate_area("rectangle", 5, 8))
print("triangle area:", calculate_area("triangle", 7, 8))
```

```python
#Task6
sentence = input("enter a sentence: ")
vowels = "aeiouAEIOU"
count = 0
for char in sentence:
    if char in vowels:
        count += 1
print("number of vowels:", count)
print("reversed sentence:", sentence[::-1])
print("uppercase sentence:", sentence.upper())
```
